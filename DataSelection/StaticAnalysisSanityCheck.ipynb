```python
import csv
import re
from pathlib import Path

# === Paths ===
bump_instances_root = Path("/Volumes/Rachna-HD/Dataset/StaticAnalysis")
report_folder = bump_instances_root / "MissingReports"
report_folder.mkdir(parents=True, exist_ok=True)
output_csv_path = report_folder / "missing_json_usage_files.csv"

# === Collect valid BUMP instances ===
bump_dirs = [d for d in bump_instances_root.iterdir() if d.is_dir() and re.match(r"^BBC\d+$", d.name)]
missing_json_instances = []

print(" Checking each BUMP instance for UsageReport JSONs...\n")

for bump_dir in bump_dirs:
    usage_dir = bump_dir / "UsageReport"

    if not usage_dir.exists():
        print(f"{bump_dir.name} ➜  'UsageReport' folder missing")
        missing_json_instances.append([bump_dir.name])
    else:
        json_files = list(usage_dir.rglob("*.json"))
        if not json_files:
            print(f"{bump_dir.name} ➜  No JSON files found in 'UsageReport'")
            missing_json_instances.append([bump_dir.name])
        else:
            print(f"{bump_dir.name} ➜  Found {len(json_files)} JSON file(s)")

# === Summary ===
print(f"\n=== SUMMARY ===")
print(f"Total BUMP instances scanned: {len(bump_dirs)}")
print(f"BUMP instances missing JSON: {len(missing_json_instances)}")

# === Save results to CSV ===
with output_csv_path.open("w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Missing_BUMP_Instance_ID"])
    writer.writerows(missing_json_instances)

print(f"\n Missing JSON usage report saved to: {output_csv_path}")
```

```python
import pandas as pd
from pathlib import Path

# === Paths ===
bump_csv_path = Path("Dataset/FinalBUMP_Instances.csv")  # Your provided path
missing_report_csv = Path("/Volumes/Rachna-HD/Dataset/StaticAnalysis/MissingReports/missing_json_usage_files.csv")
detailed_output_csv = Path("/Volumes/Rachna-HD/Dataset/StaticAnalysis/MissingReports/detailed_missing_report.csv")

# === Load data ===
missing_df = pd.read_csv(missing_report_csv, dtype=str)
bump_df = pd.read_csv(bump_csv_path, dtype=str)

# === Clean IDs and drop bad entries like 'MissingReports' just in case ===
missing_ids = (
    missing_df['Missing_BUMP_Instance_ID']
    .str.strip()
    .dropna()
    .unique()
)
missing_ids = [id for id in missing_ids if re.match(r"^BBC\d+$", id)]  # enforce valid format
bump_df['custom_id'] = bump_df['custom_id'].str.strip()

# === Filter metadata to only missing IDs ===
filtered_df = bump_df[bump_df['custom_id'].isin(missing_ids)]

# === Save detailed report ===
filtered_df.to_csv(detailed_output_csv, index=False)
print(f" Saved {len(filtered_df)} detailed missing BUMP entries to: {detailed_output_csv}")
```
